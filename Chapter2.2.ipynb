Bài 7: Một số được gọi là số Amstrong bậc N, nếu nó là số nguyên dương và tổng bậc N của các các chữ số cấu thành thì bằng chính số đó.Ví dụ: 
    Số Amstrong bậc 4:  abcd = a4 + b4 + c4 + d4
    Số Amstrong bậc N: abc….d  = aN + bN + cN + ….+ dN
    Cụ thể: 153 = 1*1*1 + 5*5*5 + 3*3*3  nên 153 số Amstrong.
Viết chương trình nhập vào 1 số, cho biết số đó có phải là số Amstrong hay không?

```python
num = int(input("Enter a number: "))
level = int(input("Bậc: "))
sums = 0
temp = num
while temp > 0:
   digit = temp % 10
   sums += digit ** level
   temp //= 10
if num == sums:
   print(num, "is Amstrong, level : " + str(level))
else:
   print(num, "is not Amstrong")
```

Bài 19: Hiển thị 10 từ có tuần suất xuất hiện nhiều nhất (phổ biến nhất)

```python
ftext = open("romeo.txt")
tu_dien_cac_tu = {}
for dong in ftext:
    danh_sach_tu = dong.split()
    for tu in danh_sach_tu:
        tu_dien_cac_tu[tu] = tu_dien_cac_tu.get(tu, 0) + 1

danh_sach = []
for key, val in tu_dien_cac_tu.items():
    newtup = (val, key) 
    danh_sach.append(newtup)

danh_sach = sorted(danh_sach, reverse=True)

for val, key in danh_sach[:10] :
    print(key, val)
```

```python
a, b, c = map(float, input("Nhập vào 3 cạnh của tam giác: ").split())
match (a, b, c):
    case (a, b, c) if a + b > c and a + c > b and b + c > a:
        if a == b == c:
            print("Tam giác đều")
        elif a == b or b == c or a == c:
            print("Tam giác cân")
        elif (a**2 == b**2 + c**2) or (b**2 == a**2 + c**2) or (c**2 == a**2 + b**2):
            print("Tam giác vuông")
        else:
            print("Tam giác thường")
    case _:
        print("Không phải tam giác")
```

Viết chương trình đọc một chuỗi từ người dùng và sử dụng một vòng lặp để xác định xem đó có phải là một Palindrom hay không. Hiển thị kết quả, bao gồm một thông báo đầu ra có ý nghĩa.

```python
chuoi_goc = input("Nhập vào 1 chuỗi")
chuoi_kiem_tra = chuoi_goc.replace(" ", "").lower()
chieu_dai = len(chuoi_kiem_tra)

chuoi_dao_nguoc = ""

for i in range(chieu_dai - 1, -1, -1):
    chuoi_dao_nguoc += chuoi_kiem_tra[i]
if chuoi_kiem_tra == chuoi_dao_nguoc:
    print(chuoi_goc, "là chuỗi Palindrome")
else:
    print(chuoi_goc, "không phải chuỗi Palindrome")
```

Bài 34: Viết chương trình chuyển đổi số thập phân thành nhị phân. Đọc số thập phân từ người dùng dưới dạng số nguyên. Hiển thị kết quả, cùng với một thông điệp thích hợp.

```python
def chuyen_doi_thap_phan_sang_nhi_phan():
    # Đọc số nguyên thập phân từ người dùng
    try:
        so_thap_phan = int(input("Nhập một số nguyên thập phân (Decimal): "))
    except ValueError:
        print("Lỗi: Đầu vào không phải là một số nguyên hợp lệ.")
        return

    # Xử lý trường hợp đặc biệt: Số 0
    if so_thap_phan == 0:
        print(f"\nKết quả: Số thập phân {so_thap_phan} chuyển thành nhị phân là: 0")
        return

    # Xử lý số âm (Optional: Nếu bạn cần xử lý số âm, bạn sẽ cần sử dụng biểu diễn bù 2)
    # Trong chương trình này, ta chỉ chuyển đổi phần giá trị tuyệt đối.
    is_am = False
    if so_thap_phan < 0:
        is_am = True
        so_chuyen_doi = abs(so_thap_phan)
    else:
        so_chuyen_doi = so_thap_phan

    chuoi_nhi_phan = ""

    # Thuật toán chuyển đổi: Chia liên tiếp cho 2 và lấy số dư
    while so_chuyen_doi > 0:
        # Lấy phần dư (Remainder): Đây là bit nhị phân (0 hoặc 1)
        phan_du = so_chuyen_doi % 2
        
        # Thêm phần dư vào đầu chuỗi nhị phân (vì chúng ta lấy từ dưới lên)
        chuoi_nhi_phan = str(phan_du) + chuoi_nhi_phan
        
        # Cập nhật số (Quotient): Lấy phần nguyên của phép chia cho 2
        so_chuyen_doi //= 2

    # Hiển thị kết quả
    if is_am:
        # Nếu là số âm, thêm dấu trừ vào trước kết quả (Lưu ý: Đây không phải biểu diễn bù 2)
        print(f"\nKết quả: Số thập phân {so_thap_phan} chuyển thành nhị phân là: -{chuoi_nhi_phan}")
        print("   (Lưu ý: Đây chỉ là giá trị tuyệt đối với dấu trừ, không phải biểu diễn bù 2.)")
    else:
        print(f"\n✅ Kết quả: Số thập phân {so_thap_phan} chuyển thành nhị phân là: {chuoi_nhi_phan}")
        print("   (Đây là kết quả sử dụng phương pháp chia liên tiếp cho 2.)")

# Chạy chương trình
chuyen_doi_thap_phan_sang_nhi_phan()
```

Bài 49: Số nguyên tố là một số nguyên lớn hơn 1 và chỉ chia hết cho 1 và chính nó. Viết hàm xác định tham số của nó có phải là số nguyên tố hay không, trả về True nếu đúng và False nếu không phải. Viết chương trình chính đọc số nguyên từ người dùng và hiển thị thông báo cho biết đó có phải là số nguyên tố hay không.

```python
n = int(input("Nhập vào 1 số :"))
def is_Prime(n):
    if n <= 1:
        return False
    else:
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                return False
        return True
if is_Prime(n):
    print(n, "là số nguyên tố")
else:
    print(n, "không phải số nguyên tố")
```

```python
import random

def tao_mat_khau_ngau_nhien():
    # Độ dài mật khẩu ngẫu nhiên từ 7 đến 10 ký tự
    do_dai = random.randint(7, 10)
    
    # Sinh ngẫu nhiên các ký tự trong khoảng ASCII từ 33 đến 126
    mat_khau = ''.join(chr(random.randint(33, 126)) for _ in range(do_dai))
    
    return mat_khau

# --- Chương trình chính ---
if __name__ == "__main__":
    mk = tao_mat_khau_ngau_nhien()
    print("Mật khẩu ngẫu nhiên được tạo:", mk)
```

Bài 51: Viết một hàm có một tham số password để xác định xem mật khẩu có tốt hay không. Một mật khẩu tốt là một mật khẩu dài ít nhất 8 ký tự và chứa ít nhất một chữ cái viết hoa, ít nhất một chữ cái viết thường và ít nhất một số. Hàm sẽ trả về True nếu mật khẩu là tốt, ngược thì nó sẽ trả về Fales. Chương trình có một chương trình chính đọc mật khẩu từ người dùng và hiển thị xem nó có tốt hay không.

```python
def kiem_tra_mat_khau(password):
    # Kiểm tra độ dài
    if len(password) < 8:
        return False
    
    # Kiểm tra có ít nhất một chữ hoa, một chữ thường, và một số
    co_chu_hoa = any(c.isupper() for c in password)
    co_chu_thuong = any(c.islower() for c in password)
    co_so = any(c.isdigit() for c in password)
    
    # Mật khẩu tốt nếu thỏa cả 3 điều kiện trên
    return co_chu_hoa and co_chu_thuong and co_so

# --- Chương trình chính ---
if __name__ == "__main__":
    mat_khau = input("Nhập mật khẩu cần kiểm tra: ")
    
    if kiem_tra_mat_khau(mat_khau):
        print("✅ Mật khẩu tốt.")
    else:
        print("❌ Mật khẩu chưa tốt.")
```

Bài 52: Một số nguyên, n, được cho là hoàn hảo khi tổng của tất cả các ước số của n bằng n. Ví dụ, 28 là một số hoàn hảo vì các ước số của nó là 1, 2, 4, 7 và 14. Tổng 1 + 2 + 4 + 7 + 14 = 28. Viết hàm xác định xem số nguyên dương có hoàn hảo hay không. 

```python
n = int(input("Nhập vào 1 số : "))
def isPerfect(n):
    sum = 0
    if n < 1:
        return False
    else:
        for i in range (1, n):
            if n % i == 0:
                sum += i
        if sum == n:
            return True
        else:
            return False
if isPerfect(n) :
    print(n, "là số hoàn hảo")
else:
    print(n, "không phải số hoàn hảo")
```

Bài 53: Viết một hàm trả về một danh sách chứa mọi danh sách con có thể có của danh sách. Ví dụ: danh sách con của [1, 2, 3] là [], [1], [2], [3], [1, 2], [2, 3] và [1, 2, 3].

```python
def tao_danh_sach_con(lst):
    ds_con = []
    n = len(lst)
    
    # Duyệt tất cả các vị trí bắt đầu và kết thúc có thể có
    for i in range(n + 1):
        for j in range(i + 1, n + 1):
            ds_con.append(lst[i:j])
    
    # Thêm danh sách rỗng
    ds_con.insert(0, [])
    return ds_con

# --- Chương trình chính ---
if __name__ == "__main__":
    danh_sach = [1, 2, 3]
    ket_qua = tao_danh_sach_con(danh_sach)
    
    print("Các danh sách con của", danh_sach, "là:")
    for sub in ket_qua:
        print(sub)
```

Bài 54: Khi viết ra một danh sách các từ bằng tiếng Anh, người ta thường phân tách các từ bằng dấu phẩy. Ngoài ra, thêm từ “and” trước từ cuối cùng, trừ khi danh sách chỉ chứa một từ. Hãy xem xét bốn danh sách sau đây: apples apples and oranges apples, oranges and bananas apples, oranges, bananas and lemons Viết hàm lấy danh sách các chuỗi làm tham số của nó. Hàm sẽ trả về một chuỗi chứa tất cả các từ trong danh sách, được định dạng theo cách được mô tả như trên. Hàm hoạt động cho danh sách có độ dài bất kỳ. Code bao gồm một chương trình chính đọc một số từ do người dùng nhập vào, định dạng chúng bằng cách gọi hàm và sau đó hiển thị kết quả mà hàm trả về

```python
def dinh_dang_danh_sach_tu(words):
    n = len(words)
    
    if n == 0:
        return ""
    elif n == 1:
        return words[0]
    elif n == 2:
        return f"{words[0]} and {words[1]}"
    else:
        # Ghép các từ đầu tiên bằng dấu phẩy, và thêm 'and' trước từ cuối
        return f"{', '.join(words[:-1])} and {words[-1]}"

# --- Chương trình chính ---
if __name__ == "__main__":
    print("Nhập danh sách các từ (nhập trống để kết thúc):")
    words = []
    
    while True:
        tu = input("Nhập từ: ").strip()
        if tu == "":
            break
        words.append(tu)
    
    ket_qua = dinh_dang_danh_sach_tu(words)
    print("\nChuỗi được định dạng:")
    print(ket_qua)
```

Bài 55. Tệp nén là một loại lưu trữ chứa một hoặc nhiều tệp đã bị giảm kích thước. Nén tập tin trong các hệ điều hành hiện đại thường khá đơn giản. Viết chương trình nén và giải nén tệp bằng Python

```python
import zipfile
import os

# --- Hàm nén nhiều tệp ---
def nen_nhieu_tap_tin(danh_sach_tap_tin, ten_zip):
    try:
        with zipfile.ZipFile(ten_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
            for ten_tap_tin in danh_sach_tap_tin:
                if os.path.isfile(ten_tap_tin):
                    zipf.write(ten_tap_tin, os.path.basename(ten_tap_tin))
                else:
                    print(f"⚠️  Bỏ qua '{ten_tap_tin}' (không phải là tệp hợp lệ)")
        print(f"✅ Đã nén {len(danh_sach_tap_tin)} tệp vào '{ten_zip}'")
    except Exception as e:
        print("❌ Lỗi khi nén tệp:", e)


# --- Hàm giải nén ---
def giai_nen_tap_tin(ten_zip, thu_muc_dich):
    try:
        with zipfile.ZipFile(ten_zip, 'r') as zipf:
            zipf.extractall(thu_muc_dich)
        print(f"✅ Đã giải nén '{ten_zip}' vào thư mục '{thu_muc_dich}'")
    except Exception as e:
        print("❌ Lỗi khi giải nén:", e)


# --- Chương trình chính ---
if __name__ == "__main__":
    print("CHƯƠNG TRÌNH NÉN VÀ GIẢI NÉN NHIỀU TỆP")
    print("1. Nén nhiều tệp")
    print("2. Giải nén tệp ZIP")

    lua_chon = input("Chọn chức năng (1 hoặc 2): ").strip()

    if lua_chon == "1":
        print("Nhập đường dẫn các tệp cần nén (nhập trống để kết thúc):")
        danh_sach_tap_tin = []
        while True:
            duongdan = input("Tệp: ").strip()
            if duongdan == "":
                break
            danh_sach_tap_tin.append(duongdan)

        if danh_sach_tap_tin:
            ten_zip = input("Nhập tên tệp ZIP muốn tạo (vd: files.zip): ").strip()
            nen_nhieu_tap_tin(danh_sach_tap_tin, ten_zip)
        else:
            print("❌ Không có tệp nào được nhập.")

    elif lua_chon == "2":
        ten_zip = input("Nhập đường dẫn tệp ZIP cần giải nén: ").strip()
        thu_muc_dich = input("Nhập thư mục đích để giải nén: ").strip()
        giai_nen_tap_tin(ten_zip, thu_muc_dich)

    else:
        print("❌ Lựa chọn không hợp lệ.")
```
